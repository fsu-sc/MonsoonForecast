```python
import xarray as xr
import numpy as np
import matplotlib.pyplot as plt
import cmocean as cmo

file_path = "tp_yearly_anomaly_florida_1979-2022.nc"
ds = xr.open_dataset(file_path)
mean = ds.attrs['climatology_average']
ds
```

```python
ds.sel(year=1990).yearly_anomaly.mean()
#plt.axhline(mean, color='red', linestyle='--', label='mean')
#plt.show()
```

```python
fig, ax = plt.subplots(dpi=300)
hov = ds.yearly_anomaly.plot(cmap=cmo.cm.balance, ax=ax, add_colorbar=False)
ax.set_xlabel('Day of Year')
ax.set_ylabel('Year')
plt.colorbar(hov, label='Anomaly (mm/day)')
#plt.axhline(mean, color='red', linestyle='--', label='mean')
ax.set_aspect('auto')
plt.tight_layout()
#plt.show()
plt.savefig('yearly_anomaly_florida_1979-2022.png', bbox_inches='tight')

```
